```python
%%sql
CREATE TABLE Bronze_Patients(
    PatientID INT,
    FirstName STRING,
    LastName STRING,
    DateOfBirth DATE,
    Gender STRING,
    City STRING,
    State STRING,
    ContactNumber STRING,
    Email STRING,
    LastModifiedDate TIMESTAMP

);
```

```python
%%sql
CREATE TABLE Bronze_Admissions (
    AdmissionID INT,
    PatientID INT ,
    AdmissionDate TIMESTAMP,
    DischargeDate TIMESTAMP,
    AdmissionReason STRING,
    Department STRING,
    DoctorName STRING,
    AdmissionStatus STRING,
    LastModifiedDate TIMESTAMP
);
```
